```python
import uproot
import awkward as ak
```

```python
threshold = 0.995
```

```python

total_counts = 0
total_sum = 0


for batch in uproot.iterate(["/scratch-cbe/users/alikaan.gueven/ML_KAAN/lgatr_outputs/predict/zjets*.root"]):
    total_counts += ak.count(batch.label_llp[batch.score_label_llp > threshold])
    total_sum += ak.sum(batch.label_llp[batch.score_label_llp > threshold])

# total_sum / total_counts
print('sig: ', total_sum)
print('bkg: ', total_counts - total_sum)
```
